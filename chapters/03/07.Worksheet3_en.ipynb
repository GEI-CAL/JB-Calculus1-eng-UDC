# Worksheet 3: Differential Calculus of Functions of One Variable

## Basic exercises

1. Let $f$ be the function given by $f(x)=5x^2.$

$~~~~~~$a) Use the definition of the derivative to show that $f'(x)=10x.$
    
$~~~~~~$b) Compute the equation of the tangent line to the graph of $f$ at $\displaystyle{x=\frac12}.$

2. Let $f : \mathbb{R} \longrightarrow \mathbb{R}$ be given by:
\begin{equation*}
f(x) =
\begin{cases}
2x - 1 & \quad \text{if } x \leq -1, \\
(x+1)^3 + 2x & \quad \text{if } x > -1.
\end{cases}
\end{equation*}
Determine whether it is differentiable at $x = -1$.

3. Let $f$ be the function given by
\begin{equation*}
f(x)=
\begin{cases}
\displaystyle{x+x^2 \sin\left(\frac{1}{x}\right)} & \text{if } x\neq 0, \\
0 & \text{if } x = 0.
\end{cases}
\end{equation*}

$~~~~~~$a) Obtain the expression of $f'(x)$ for $x\neq 0$.

$~~~~~~$b) Compute $f'(0)$ using the definition of the derivative; that is, as the limit of the difference quotient.

$~~~~~~$c) Check that the value obtained in part (b) does not coincide with the result of substituting $x=0$ into the expression found in (a). Explain why these different results occur.


4. Consider the function $f$ given by:
\begin{equation*}
f(x) =
\begin{cases}
\displaystyle{\frac{1-e^{-x}}{x}} & \quad \text{if } x \neq 0, \\
1 & \quad \text{if } x = 0.
\end{cases}
\end{equation*}

$~~~~~~$a) Compute $f'(x)$ at every point where it exists.

$~~~~~~$b) Compute $\displaystyle{\lim_{x\to -\infty}{f(x)}}$, $\displaystyle{\lim_{x\to \infty}{f(x)}}$, and the set $\textrm{Im}(f)$.

5. Write the Newton-Raphson method to approximate $\sqrt[3]{2}$ starting from $x_0 = 1$. Compute the first two iterates.


6. To approximate the value of the solution of the equation $x^3+2x-2=0$ in the interval $[-2,\, 2]$, we are going to use the Newton-Raphson method.

$~~~~~~$a) State the Newton-Raphson algorithm for this case.

$~~~~~~$b) Compute the first two iterations using that algorithm. To do this, use the value $x_0=0$ as the initial approximation.

7. Let $f(x) = \displaystyle \arctan \left( \frac{\sin(x)}{1 + \cos(x)} \right )$. Compute the tangent line to the graph of $f$ at $x = 0$.


8. Let the function $f$ be given by $f(x) = (4x+1)^{(2+ \sin(x^2))}$. Compute its derivative at any point.


9. If we consider the function $f:\mathbb{R}\rightarrow \mathbb{R}$ given by
\begin{equation*}
f(x)=\displaystyle{\left\{ \begin{array}{ccl}
\displaystyle{\frac{\sin^2(3x)}{3x^2}} & \text{if} & x\neq 0, \\
\displaystyle{\frac{1}{3}} & \text{if} & x=0,
\end{array}
\right.}
\end{equation*}
choose the correct option:

$~~~~~~$a) $\displaystyle\lim_{x\to 0}f(x)=\frac{1}{3}$.

$~~~~~~$b) It is not possible to compute $\displaystyle{\lim_{x\to 0}{f(x)}}$.

$~~~~~~$c) $\displaystyle{\lim_{x\to 0}{f(x)}}=3$.

$~~~~~~$d) It is continuous on $\mathbb{R}$.

10. Compute the values of $a$ and $b$ such that

\begin{equation*}
\displaystyle{\lim_{x\to 0}\frac{ax^2+bx+1-e^{2x}}{\sin(x^2)}=1}.
\end{equation*}


11. Let the function $f$ be given by $\displaystyle f(x) = \frac{1}{2} x |x|$. What is the class of $f$?


12. We consider the function $f$ given by
\begin{equation*}
f(x)=
\begin{cases}
\sin(x) \, \qquad \text{if } x \in (-\infty,0), \\
\arctan(x) \, \qquad \text{if } x \in [0,1], \\
\pi x^3 / 4 \, \qquad \text{if } x \in (1,+\infty).
\end{cases}
\end{equation*}

$~~~~~~$a) Sketch the graph of $f$.

$~~~~~~$b) Determine the class of $f$ on $\mathbb{R}$.

$~~~~~~$c) Compute, if possible, the tangent line to $f$ at the point $x_0 = 0$.

$~~~~~~$d) Determine a set of real numbers on which $f$ is of class $C^{\infty}$.

13. Let $f$ be the function given by
\begin{equation*}
f(x)=
\begin{cases}
\sqrt{x} + 1 & \quad \text{if } x \in (0,1), \\
\alpha x^2 + \beta x + 1 & \quad \text{if } x \in [1,2).
\end{cases}
\end{equation*}

$~~~~~~$a) Determine $\alpha$ and $\beta$ so that $f$ is differentiable on $(0,2)$.

$~~~~~~$b) What conditions must $\alpha$ and $\beta$ satisfy so that $f$ has a relative minimum at the point $x=1$?

14. Compute the relative extrema, and the absolute extrema if they exist, of the function defined on the interval
$\left[-\frac{1}{2},\frac{3}{2}\right]$ by $f(x) = x^2 - 2|x| + 2$.


15. Consider the function $f:\mathbb{R} \rightarrow \mathbb{R}$ given by $\displaystyle{f(x)=\frac{|x|}{e^{x-1}}}$.

$~~~~~~$a) Study the continuity and differentiability of the function.

$~~~~~~$b) Sketch the graph of $f$. To do this:
- compute its relative extrema, inflection points, and asymptotes.
- Also determine its interval(s) of concavity and/or convexity.

$~~~~~~$c) Determine the absolute extrema of $f$.


16. Show that the equation $x^4-4x^3-1=0$ has a unique root in the interval $[4,5]$.
State the Newton-Raphson method to solve the equation from the previous part. Starting from $x_0=4$, obtain the approximation $x_1$ by hand and the approximation $x_7$ using `Python`.


17. On what interval is the function $f$ given by $f(x)=x^3e^x$ increasing? Is it concave or convex on that interval? Sketch its graph.


18. Find the condition that $\lambda$ must satisfy so that the polynomial $x^4 + x^3 + \lambda x^2$ is concave on some interval. Determine that interval in terms of $\lambda$.


19. Find a cubic polynomial with no local extrema but with an inflection point and a horizontal tangent at $P=(1,\,20)$.


20. Sketch the graph of a function whose slope is always positive and increasing. Do you know any elementary function that is an example of this situation? <br> Sketch the graph of a function whose slope is always positive and decreasing. Do you know any elementary function that is an example of this situation?


21. A rectangle whose base lies on the $x$-axis has its two upper vertices on the parabola $y = 12 - x^2$. What is the largest area that this rectangle can have? Indicate its dimensions.


22. Knowing that $p(x)=5+(x+1)^2+3(x+1)^3$ is the third-order Taylor polynomial centered at $x_0=-1$ of a function $f$, compute $f(-1)$, $f'(-1)$, and $f''(-1)$. Justify your answer.


23. We consider the function $f$ defined by $f(x)=\ln(1+x)$.

$~~~~~~$a) Compute the McLaurin polynomial (that is, the Taylor polynomial with $x_0=0$) of order two for $f$.

$~~~~~~$b) Use this polynomial to approximate the value of $\ln(1.1)$, bounding the error made.

24. Construct the first-order Taylor polynomial, $p$, for the function $g(x)=\sin(x)$, centered at the point $x_0=\pi/2$.
Now consider the function $f$ defined as follows:
\begin{equation*}
f(x)=
\begin{cases}
\sin(x) \quad & \quad \text{if } x \leq \pi/2, \\
p(x) \quad & \quad \text{if } x > \pi/2,
\end{cases}
\end{equation*}
where $p$ is the polynomial constructed in the previous part. What is the class of $f$ on $\mathbb{R}$?


25. Determine the second-order Taylor polynomial of the function $f(x)=\arcsin(x)$ at the point $x_0=0$. <br> Use the previous polynomial to obtain, approximately, the angle whose sine is $\frac{1}{10}$.


26. Let the function be given by
\begin{equation*}
\begin{array}{cccl}
f: & \mathcal{D}om(f) \subset \mathbb{R} & \longrightarrow & \mathbb{R} \\
& x & \rightsquigarrow & f(x)=\dfrac{x^2}{e^x}
\end{array}
\end{equation*}

$~~~~~~$a) Compute the domain of $f$.

$~~~~~~$b) Compute the absolute extrema of $f$ on the interval $[0,10]$, first justifying their existence.

$~~~~~~$c) Compute the absolute extrema of $f$ on its domain.
