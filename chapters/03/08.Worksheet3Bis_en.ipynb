## Exercises from previous years' third Python lab test


**[1]** Consider the function $f$ defined by $f(x) = \left|\log(x)\right| - \frac{1}{2}$ on the interval $I=[\frac{1}{2},\frac{3}{2}]$.

* Plot it using `Matplotlib`.
* Use `Python` to check that it has a root in $I$.
* For extra credit: can you prove that the root is unique?
* Approximate it using Newton-Raphson with $x_{0}=\frac{8}{10}$ and relative error less than $10^{-9}$.
* Approximate it using Newton-Raphson with $x_{0}=\frac{12}{10}$ and relative error less than $10^{-9}$.
* Can you explain what happened in the two previous parts?
* Use `Python` to calculate the absolute maximum and minimum of $f$ on $I$.

**[3]** Consider the equation $\sin(3x)=0$.

* Prove that it has a unique root in the interval $[\pi/6,\pi/2]$.
* Use Newton's method to approximate a root of the equation with tolerance less than $10^{-10}$, starting from $x_0=1$ and from $x_0=\pi/6$.

The code should display the function values at each successive approximation. Interpret the output.
