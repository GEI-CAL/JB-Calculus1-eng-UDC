## Exercises from previous years' second Python lab test


**[1]** Consider the function

$$
f(x) = \frac{\ln(\tan(x)) +x}{x^2 - 2} - 1
$$

* Check that this function has a root in the interval $[0,1]$.
* Use the bisection method to approximate this root with an absolute error less than
  * $10^{-1}$
  * $10^{-5}$
  * $10^{-12}$
* Try to apply the bisection method to approximate this root in the interval $[-1.6,1]$. Interpret the result.
* Relate what you have just done to finding a solution in the interval $[0,0.5]$ of the equation
  $$
  \ln(\tan(x)) + x = \frac{1}{2} \left(x^2 - 2\right).
  $$
* Use `Matplotlib` to plot $f$ on the interval $[x_1,1]$, where $x_1$ is the first approximation obtained in the first part.

**[5]** Consider the function $f: [-2,2] \to \mathbb{R}$ given by $f(x)=x^3+x+1$.

**[6]** Consider the function

$$
f(x) = \tan(x-1)+5
$$

and the points $x_0= -0.50$, $x_1 = -0.45$, $x_2 = -0.30$, $x_3 = 0.30$, $x_4 = 0.70$, $x_5 = 1$, $x_6 = 1.2$, $x_7 = 1.40$, $x_8=1.80$, $x_9=2.10$ and $x_{10} = 2.40$.

* Use `NumPy` to construct a polynomial whose value is $5$ at $x_3$ and $0$ at the other nodes. Call this polynomial $L_3$.
* Use `NumPy` and divided differences to construct the Lagrange interpolation polynomial $p_{10}$ passing through the eleven selected points.
* Use $p_{10}$ to approximate $B = \tan(2) + 6$.
* Use `Matplotlib` to plot $f$, $L_3$, and $p_{10}$ on the interval $[x_0,x_{10}]$.


**[7]** We want to approximate the function

$$
f(x) = \frac{\cos(x)+\sin(x)}{\cos(x)+2}
$$

on the interval $[-6,6]$ with a Lagrange polynomial. Divide the interval into $n$ equally spaced points, evaluate the function at those points, and compute the corresponding Lagrange polynomial. For example, if $n=13$, the points used are $x_0=-6$, $x_1=-5,\ldots,x_{12}=6$.

1. For $n=13$, use `NumPy` and divided differences to construct the corresponding Lagrange interpolation polynomial. Use it to approximate $f(0.30)$ and calculate the relative error, as a percentage,
   $$
   \frac{|p_{12}(0.30) - f(0.30)|}{|f(0.30)|}100.
   $$
   What is the same error at $-5.5$?
2. Repeat the experiment with $n=25$ and $n=40$ at both points, $0.30$ and $-5.5$. Interpret the results.
3. Use `Matplotlib` to plot $f$, $p_{12}$, $p_{24}$, and $p_{39}$ on the interval $[-6,6]$.


**[8]** Let $s$ be the function defined by $s(x) = \dfrac {e^x}{e^x + 1} - 0.6$.

1. Construct a vector $x$ of 9 equally spaced points, including the endpoints of $[-2,2]$.
2. Let $p$ be the Lagrange polynomial interpolating the table formed by those points and their images under $s$. Use divided differences to check $p(1)$.
3. Use $p$ to estimate the value at $1.25$ and calculate the absolute error relative to $s(1.25)$. Give at least 6 decimal places in parts 2 and 3.


**[9]** The winning jockey's passage times at selected points on a 2 km racecourse are given below:

* Distance markers (km): $[0,\;0.5,\;1,\;1.5,\;2.0]$.
* Passage times (s): $[0,\;50,\;90,\;130,\;195]$.

1. Construct the corresponding divided-difference table. Show only the divided differences used to construct the interpolation polynomial.
2. Check that the polynomial interpolates all the given data. Use the Lagrange polynomial based on divided differences to estimate the passage time at 1.25 km.
3. Repeat the process with the data order reversed, expressing the distance markers as a function of passage time. If you estimate the distance corresponding to the time found in part 2, do you obtain 1.25 km? Justify your answer.
