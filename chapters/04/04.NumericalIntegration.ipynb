# Numerical integration


## Cases where it is used

There are some cases in which, instead of looking for an antiderivative of a function and applying Barrow's rule, we use a numerical integration formula. In particular, when

* we only know the values of the function at a finite number of points, 
* its antiderivative cannot be expressed in terms of elementary functions. For example:

$$ 
\int \frac{\sin x}{x} \,dx;\qquad \int e^{-x^2}\,dx.
$$
* its antiderivative is very expensive to compute or evaluate. For example:

$$
\int \frac{1}{(x-8)\sqrt{x^2-4x-7}} \, dx.
$$

````{prf:definition} Numerical integration or quadrature formula
:label: def_intnum 
:nonumber: 
A **numerical integration** formula, or **quadrature** formula, is a sum of the form:

$$
\sum_{i=0}^n\omega_i\,f(x_i)\approx \int_a^bf(x)\,dx,
$$

where the points $x_0$, $x_1$, ..., $x_n$ are called quadrature nodes and the values $\omega_0$, $\omega_1$, ..., $\omega_n$ are the weights associated with each node.
````


## Simple formulas

* **Midpoint formula**:
    <center>

    $$
    \int_a^bf(x)\,dx\,\simeq\,(b-a)\,f\left(\frac{a+b}{2}\right) \,.
    $$
    <img src="../../images/cap4_midpoint.png" width="300"/>
    </center>

* **Trapezoidal formula**:
    <center>

    $$
    \int_a^bf(x)\,dx\,\simeq\,\frac{b-a}{2}\,\big(f(a) + f(b)\big) 
    $$
    <img src="../../images/cap4_trapezoid.png" width="300"/>
    </center>

* **Simpson's formula**: It is based on quadratic interpolation (three nodes):
    <center>

    $$
    \int_a^bf(x)\,dx
    \,\simeq\,
    \frac{b-a}{6}\,\Big(\,f(a)\,+\,4\,\, f(\frac{a+b}{2})\,+\,f(b)\Big)
    $$
    <img src="../../images/cap4_simpson.png" width="300"/>
    </center>

````{prf:example} 
:label: ex_intnumersimple 
:nonumber: 
Find an approximation to $\displaystyle \int_1^3 \sin(\sqrt{x}) \, dx$ using the trapezoidal and Simpson formulas, and compare it with the exact solution:

Taking the substitution $u = x^{1/2}$ and integrating by parts, we obtain:

$$
\displaystyle \int \sin(\sqrt{x}) \, dx = 2 \sin(\sqrt{x}) - 2 \sqrt{x} \cos(\sqrt{x}) + C; \quad \displaystyle \int_1^3 \sin(\sqrt{x}) \, dx = 1.9279...
$$
Whereas numerical integration gives:

\begin{eqnarray*}
&\bullet& \quad \text{Trapezoid: }\displaystyle \int_1^3 \sin(\sqrt{x}) \simeq \frac{3 - 1}{2}\big(\sin(1) + \sin(\sqrt{3}) \big) = 1.8285...\\
&\bullet& \quad \text{Simpson: } \displaystyle \int_1^3 \sin(\sqrt{x}) \simeq \frac{3 - 1}{6}\big(\sin(1) + 4 \sin(\sqrt{2}) + \sin(\sqrt{3}) \big) = 1.9265...
\end{eqnarray*}

Simpson's formula uses one more node than the trapezoidal rule, and in this example we can see that it provides a better approximation.
````


## Composite formulas

The idea is simple: divide the interval of integration, $[a,b]$, into subintervals and use a simple numerical integration formula on each of them.

The most common case arises when we take $n$ subintervals, $[x_i, x_{i + 1}]$, all of equal length $h$. That is, when we choose
$x_i = a + ih$, $i = 0, 1, ..., n$, for a value $h = \frac{b - a}{n}$.

Then we approximate the integral by a simple formula on each subinterval. But be careful! 
* for the midpoint or Simpson rule we must take intervals of size $2h$, that is, every 3 points, because we need the midpoint of each subinterval, $[x_{2i-2}, x_{2i}]$,
* for the trapezoidal rule, since we do not need the midpoint, we can take all the subintervals, which in this case have length $h$: $[x_{i-1}, x_{i}]$.

Then we have:

* Composite midpoint or Simpson, with even $n$: $\displaystyle \int_a^b f(x)\,dx = \sum_{i=1}^{n/2} \int_{x_{2i-2}}^{x_{2i}} f(x)\,dx$.
* Composite trapezoidal rule: $\displaystyle \int_a^b f(x)\,dx = \sum_{i=1}^{n} \int_{x_{i-1}}^{x_{i}} f(x)\,dx$.

Now, depending on the simple formula we choose, we obtain:

* **Composite midpoint formula**:
    <center>

    \begin{eqnarray*}
    \int_a^b f(x)\,dx = \sum_{i=1}^{n/2} \int_{x_{2i-2}}^{x_{2i}} f(x)\, dx &\approx& \sum_{i=1}^{n/2} \left(x_{2i}-x_{2i-2}\right) f\left(x_{2i-1}\right) \\
    &=& 2h\sum_{i=1}^{n/2} f\left(x_{2i-1}\right)
    \end{eqnarray*}
    <img src="../../images/cap4_midpoint_composite.png" width="300"/>
    </center>

* **Composite trapezoidal formula**:
    <center>

    \begin{eqnarray*}
    \int_a^b f(x)\,dx = \sum_{i=1}^{n} \int_{x_{i-1}}^{x_{i}} f(x)\, dx &\approx& 
    \sum_{i=1}^{n} \frac{h}{2} \left( f\left(x_{i-1}\right) + f\left(x_{i}\right) \right) \\
    &=& \frac{h}{2} \left( f(x_{0}) + 2 \sum_{i=1}^{n-1}f(x_{i}) + f(x_{n}) \right)
    \end{eqnarray*}
    <img src="../../images/cap4_trapezoid_composite.png" width="300"/>
    </center>

* **Composite Simpson's formula**:
    <center>

    \begin{eqnarray*}
    \int_a^b f(x)\,dx = \sum_{i=1}^{n/2} \int_{x_{2i-2}}^{x_{2i}} f(x)\, dx &\approx& \sum_{i=1}^{n/2} \frac{2h}{6} \left(f\left(x_{2i-2}\right) + 4f\left(x_{2i-1}\right) + f\left(x_{2i}\right) \right) \\
    &=& \frac{2h}{6} \left( f(x_{0}) + 4\sum_{i=1}^{n/2} f(x_{2i-1}) + 2\sum_{i=1}^{n/2-1}f(x_{2i}) + f(x_{n}) \right)
    \end{eqnarray*}
    <img src="../../images/cap4_simpson_composite.png" width="300"/>
    </center>

## Programming in `Numpy`

The following examples implement simple and composite numerical integration formulas in `Numpy`, using $I=\int_0^3(x^4+1)\,dx$.


### Simple formulas

Below are functions that implement the simple formulas we have just studied, followed by an example of their application in `Numpy`.

We will use

$$
I=\int_{0}^{3}\left(x^4+1\right)\,dx,
$$
since in this simple case we can calculate the exact value of the integral:

$$
I=\int_{0}^{3}\left(x^4+1\right)\,dx = \left[\frac{x^5}{5}+x\right]_{x=0}^{3} = \frac{3^5}{5}+3 = 51.6\,.
$$

`````python
import sympy as sp
import numpy as np

def midpoint(a, b, fpm):
    return (b - a) * fpm

def trapezoid(a, b, fa, fb):
    return (b - a) * (fa + fb) / 2

def simpson(a, b, fa, fpm, fb):
    return (b - a) * (fa + 4 * fpm + fb) / 6

x = sp.Symbol('x', real=True)
f_expr = x**4 + 1
f = sp.lambdify(x, f_expr)

a = 0
b = 3
pm = (a + b) / 2

fa = f(a)
fpm = f(pm)
fb = f(b)

print('Approximation using the midpoint formula = ', midpoint(a, b, fpm))
print('Approximation using the trapezoidal formula = ', trapezoid(a, b, fa, fb))
print("Approximation using Simpson's formula = ", simpson(a, b, fa, fpm, fb))
`````

### Composite formulas

As you can see in the previous section, simple formulas can produce rather poor results. We will now efficiently implement the composite formulas using the `np.sum` function.

`````python
import sympy as sp
import numpy as np

x = sp.Symbol('x', real=True)
f_expr = x**4 + 1
f = sp.lambdify(x, f_expr)

a = 0
b = 3
n = 100

x1 = np.linspace(a, b, n + 1)  # Store x_0, x_1, ..., x_n
y1 = f(x1)

h = (b - a) / n  # Subinterval width

approx_trapezoid = h / 2 * (y1[0] + 2 * np.sum(y1[1:n]) + y1[n])
approx_midpoint = 2 * h * np.sum(y1[1:n:2])
approx_simpson = 2 * h / 6 * (y1[0] + 4 * np.sum(y1[1:n:2]) + 2 * np.sum(y1[2:n-1:2]) + y1[n])

print('approx_trapezoid: ', approx_trapezoid)
print('approx_midpoint: ', approx_midpoint)
print('approx_simpson: ', approx_simpson)
print('Exact: ', b**5 / 5 + b)
`````
